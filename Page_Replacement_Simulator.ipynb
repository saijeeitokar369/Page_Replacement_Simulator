# **FIFO (First In First Out) Page Replacement**

```python
from collections import deque

reference_string = [1, 2, 3, 1, 4, 2, 5]

num_frames = 3
frames = deque()

page_faults = 0
page_hits = 0

print(f"{'Ref':<5}{'Frames':<15}{'Result'}")
print("-" * 30)

for page in reference_string:
    if page in frames:
        page_hits += 1
        result = "Hit"
    else:
        page_faults += 1
        if len(frames) == num_frames:
            frames.popleft()      # remove the oldest page
        frames.append(page)
        result = "Fault"

    print(f"{page:<5}{str(list(frames)):<15}{result}")

print("-" * 30)
print(f"Total Page Faults : {page_faults}")
print(f"Total Page Hits   : {page_hits}")
print(f"Hit Ratio         : {page_hits / len(reference_string):.2f}")
print(f"Fault Ratio       : {page_faults / len(reference_string):.2f}")
```

# **LRU (Least Recently Used)**

```python
reference_string = [1, 2, 3, 1, 4, 2, 5]
num_frames = 3

frames = deque()
page_faults = 0
page_hits = 0
print(f"{'Ref':<5}{'Frames':<15}{'Result'}")
print("-" * 30)
for page in reference_string:
    if page in frames:
        page_hits += 1
        frames.remove(page)
        frames.append(page)
        result = "Hit"
    else:
        page_faults += 1
        if len(frames) == num_frames:
            frames.popleft()      # remove the oldest page
        frames.append(page)
        result = "Fault"
    print(f"{page:<5}{str(list(frames)):<15}{result}")
print("-" * 30)
print(f"Total Page Faults : {page_faults}")
print(f"Total Page Hits   : {page_hits}")
print(f"Hit Ratio         : {page_hits / len(reference_string):.2f}")
print(f"Fault Ratio       : {page_faults / len(reference_string):.2f}")
```

# **Optimal Page Replacement**

```python
reference_string = [1, 2, 3, 1, 4, 2, 5]
num_frames = 3

frames = []
page_faults = 0
page_hits = 0

print(f"{'Ref':<5}{'Frames':<15}{'Result'}")
print("-" * 30)

for i, page in enumerate(reference_string):

    if page in frames:
        page_hits += 1
        result = "Hit"

    else:
        page_faults += 1

        if len(frames) == num_frames:

            future = reference_string[i + 1:]

            future_use = []

            for frame_page in frames:
                if frame_page in future:
                    future_use.append(future.index(frame_page))
                else:
                    future_use.append(float('inf'))

            replace_index = future_use.index(max(future_use))

            frames[replace_index] = page

        else:
            frames.append(page)

        result = "Fault"

    print(f"{page:<5}{str(frames):<15}{result}")

print("-" * 30)
print(f"Total Page Faults : {page_faults}")
print(f"Total Page Hits   : {page_hits}")
print(f"Hit Ratio         : {page_hits / len(reference_string):.2f}")
print(f"Fault Ratio       : {page_faults / len(reference_string):.2f}")
```
